The sales price per unit is calculated by dividing the total daily sales by the quantity sold. This helps determine the average selling price of one unit for each day.

# 4. Calculate Daily Gross Profit Percentage

Gross profit is calculated by subtracting the cost of sales from total sales. The gross profit percentage shows how much profit was made from each day's sales.

```python
%sql

SELECT
    Date,
    Sales,
    `Cost Of Sales`,
    ROUND(((Sales-`Cost Of Sales`)/Sales)*100,2) AS Gross_Profit_Percentage
FROM dbacademy.default.sales_case_study
ORDER BY Date;
```

# 7. Promotional Periods and Price Elasticity of Demand

## Objective
The purpose of this section is to identify promotional periods where the unit price decreased while the quantity sold increased. These periods are used to calculate the Price Elasticity of Demand (PED) and understand how customers respond to price changes.

```python
%sql
-- This query compares each day's unit price and quantity sold with the previous day.
-- It helps identify promotional periods where prices decreased and sales increased.
SELECT
    Date,
    ROUND(Sales / `Quantity Sold`, 2) AS Unit_Price,
    `Quantity Sold`,
    LAG(ROUND(Sales / `Quantity Sold`, 2))
        OVER (ORDER BY Date) AS Previous_Unit_Price,
    LAG(`Quantity Sold`)
        OVER (ORDER BY Date) AS Previous_Quantity
FROM dbacademy.default.sales_case_study
ORDER BY Date;
```

# 7.1 Identify Promotional Periods

## Objective
This query identifies promotional periods where the unit price decreased while the quantity sold increased compared to the previous day. These periods will be used to calculate the Price Elasticity of Demand (PED).

```python
%sql
-- This query automatically finds promotional periods.
-- A promotional period is identified when the unit price decreases
-- while the quantity sold increases compared to the previous day.

WITH SalesComparison AS (

SELECT
    Date,
    ROUND(Sales / `Quantity Sold`,2) AS Unit_Price,
    `Quantity Sold`,
    LAG(ROUND(Sales / `Quantity Sold`,2))
        OVER(ORDER BY Date) AS Previous_Unit_Price,
    LAG(`Quantity Sold`)
        OVER(ORDER BY Date) AS Previous_Quantity

FROM dbacademy.default.sales_case_study

)

SELECT *
FROM SalesComparison
WHERE Unit_Price < Previous_Unit_Price
AND `Quantity Sold` > Previous_Quantity
ORDER BY Date;
```

# 8. Price Elasticity of Demand (PED)

## Objective
This section calculates the Price Elasticity of Demand (PED) for selected promotional periods. PED measures how much customer demand changes when the product price changes.

```python
%sql
-- This query calculates the Price Elasticity of Demand (PED)
-- for the first three promotional periods identified.
-- PED measures how customer demand changes when prices change.

WITH SalesComparison AS (

SELECT
    Date,
    ROUND(Sales / `Quantity Sold`,2) AS Unit_Price,
    `Quantity Sold`,
    LAG(ROUND(Sales / `Quantity Sold`,2))
        OVER(ORDER BY Date) AS Previous_Unit_Price,
    LAG(`Quantity Sold`)
        OVER(ORDER BY Date) AS Previous_Quantity

FROM dbacademy.default.sales_case_study

)

SELECT
    Date,
    Unit_Price,
    Previous_Unit_Price,
    `Quantity Sold`,
    Previous_Quantity,

    ROUND(
        (
            (`Quantity Sold` - Previous_Quantity) / Previous_Quantity
        )
        /
        (
            (Unit_Price - Previous_Unit_Price) / Previous_Unit_Price
        ),
        2
    ) AS PED

FROM SalesComparison

WHERE Unit_Price < Previous_Unit_Price
AND `Quantity Sold` > Previous_Quantity

LIMIT 3;
```
